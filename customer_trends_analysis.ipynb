```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Optional: Set plot style
plt.style.use("ggplot")

# Step 2: Load the dataset
df = pd.read_csv("shopping_trends_updated.csv")

# Step 3: Display first 5 rows
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Step 4: Display dataset shape
print("\nDataset Shape:")
print(df.shape)
```

```python
# Step 5: Display column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
#Step 6: Display data types
print("\nData Types:")
print(df.dtypes)
```

```python
# Step 7: Display dataset information
print("\nDataset Information:")
df.info()
```

```python
# ==========================================================
# Step 2: Data Cleaning & Quality Assessment
# ==========================================================

# 1. Check for missing values
print("=" * 60)
print("Missing Values")
print("=" * 60)
print(df.isnull().sum())



```

```python
# 2. Check for duplicate records
print("\n" + "=" * 60)
print("Duplicate Records")
print("=" * 60)
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
# 3. Check unique values in each column
print("\n" + "=" * 60)
print("Unique Values Count")
print("=" * 60)
print(df.nunique())
```

```python
# 4. Display summary statistics for numerical columns
print("\n" + "=" * 60)
print("Numerical Summary Statistics")
print("=" * 60)
display(df.describe())
```

```python
# 5. Display summary statistics for categorical columns
print("\n" + "=" * 60)
print("Categorical Summary Statistics")
print("=" * 60)
display(df.describe(include='object'))
```

```python
# 6. Verify data types
print("\n" + "=" * 60)
print("Data Types")
print("=" * 60)
print(df.dtypes)
```

```python
# 7. Rename columns to Python-friendly snake_case
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("(", "", regex=False)
      .str.replace(")", "", regex=False)
)
```

```python
# 8. Display updated column names
print("\n" + "=" * 60)
print("Updated Column Names")
print("=" * 60)
print(df.columns.tolist())
```

```python
# 9. Verify dataset shape after cleaning
print("\n" + "=" * 60)
print("Dataset Shape After Cleaning")
print("=" * 60)
print(df.shape)

print("\n✅ Data Cleaning & Quality Assessment Completed Successfully!")
```

```python
# ==========================================================
# Step 3: Exploratory Data Analysis (EDA)
# ==========================================================

# 1. Total Purchase Amount
print("=" * 60)
print("Total Purchase Amount")
print("=" * 60)
print(f"${df['purchase_amount_usd'].sum():,.2f}")
```

```python
# 2. Average Purchase Amount
print("\n" + "=" * 60)
print("Average Purchase Amount")
print("=" * 60)
print(f"${df['purchase_amount_usd'].mean():.2f}")
```

```python
# 3. Total Purchase Amount by Category
print("\n" + "=" * 60)
print("Purchase Amount by Category")
print("=" * 60)
display(
    df.groupby("category")["purchase_amount_usd"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# 4. Average Purchase Amount by Category
print("\n" + "=" * 60)
print("Average Purchase Amount by Category")
print("=" * 60)
display(
    df.groupby("category")["purchase_amount_usd"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# 5. Purchase Amount by Gender
print("\n" + "=" * 60)
print("Purchase Amount by Gender")
print("=" * 60)
display(
    df.groupby("gender")["purchase_amount_usd"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# 6. Purchase Amount by Season
print("\n" + "=" * 60)
print("Purchase Amount by Season")
print("=" * 60)
display(
    df.groupby("season")["purchase_amount_usd"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# 7. Top 10 Locations by Purchase Amount
print("\n" + "=" * 60)
print("Top 10 Locations by Purchase Amount")
print("=" * 60)
display(
    df.groupby("location")["purchase_amount_usd"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)
```

```python
# 8. Average Review Rating by Category
print("\n" + "=" * 60)
print("Average Review Rating by Category")
print("=" * 60)
display(
    df.groupby("category")["review_rating"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)
```

```python
# 9. Subscription Status Distribution
print("\n" + "=" * 60)
print("Subscription Status Distribution")
print("=" * 60)
display(df["subscription_status"].value_counts().reset_index())
```

```python
# 10. Payment Method Distribution
print("\n" + "=" * 60)
print("Payment Method Distribution")
print("=" * 60)
display(df["payment_method"].value_counts().reset_index())
```

```python
# 11. Discount Applied Distribution
print("\n" + "=" * 60)
print("Discount Applied Distribution")
print("=" * 60)
display(df["discount_applied"].value_counts().reset_index())
```

```python
# 12. Promo Code Used Distribution
print("\n" + "=" * 60)
print("Promo Code Used Distribution")
print("=" * 60)
display(df["promo_code_used"].value_counts().reset_index())

print("\n✅ Exploratory Data Analysis (EDA) Completed Successfully!")
```

```python
# ==========================================================
# Step 4: Data Visualization
# ==========================================================

import matplotlib.pyplot as plt

# 1. Purchase Amount by Category
plt.figure(figsize=(8,5))
df.groupby("category")["purchase_amount_usd"].sum().sort_values().plot(kind="bar")
plt.title("Total Purchase Amount by Category")
plt.xlabel("Category")
plt.ylabel("Purchase Amount (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Purchase Amount by Gender
plt.figure(figsize=(6,4))
df.groupby("gender")["purchase_amount_usd"].sum().plot(kind="bar")
plt.title("Total Purchase Amount by Gender")
plt.xlabel("Gender")
plt.ylabel("Purchase Amount (USD)")
plt.tight_layout()
plt.show()
```

```python
# 3. Purchase Amount by Season
plt.figure(figsize=(6,4))
df.groupby("season")["purchase_amount_usd"].sum().plot(kind="bar")
plt.title("Total Purchase Amount by Season")
plt.xlabel("Season")
plt.ylabel("Purchase Amount (USD)")
plt.tight_layout()
plt.show()

```

```python
# 4. Top 10 Locations by Purchase Amount
plt.figure(figsize=(10,5))
df.groupby("location")["purchase_amount_usd"].sum().sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("Top 10 Locations by Purchase Amount")
plt.xlabel("Location")
plt.ylabel("Purchase Amount (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 5. Payment Method Distribution
plt.figure(figsize=(7,7))
df["payment_method"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Payment Method Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()

```

```python
# 6. Subscription Status Distribution
plt.figure(figsize=(5,5))
df["subscription_status"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Subscription Status")
plt.ylabel("")
plt.tight_layout()
plt.show()

```

```python
# 7. Purchase Amount Distribution
plt.figure(figsize=(8,5))
plt.hist(df["purchase_amount_usd"], bins=10)
plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Frequency")
plt.tight_layout()
```

```python
# 8. Age Distribution
plt.figure(figsize=(8,5))
plt.hist(df["age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()

```

```python
# 9. Review Rating Distribution
plt.figure(figsize=(8,5))
plt.hist(df["review_rating"], bins=10)
plt.title("Review Rating Distribution")
plt.xlabel("Review Rating")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

```

```python
# 10. Previous Purchases Distribution
plt.figure(figsize=(8,5))
plt.hist(df["previous_purchases"], bins=10)
plt.title("Previous Purchases Distribution")
plt.xlabel("Previous Purchases")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()

print("\n✅ Data Visualization Completed Successfully!")
```
